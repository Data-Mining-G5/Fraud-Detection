## Role 1: Data Preprocessing

Responsible: John Cha

Script used: `preprocess_data.py`

This role loads the dataset from `archive.zip`, performs duplicate/null/class-balance checks, creates stratified train/validation/test splits, scales `Time` and `Amount` with the scaler fit on train only, and writes the shared artifacts consumed by later roles.

```python
from preprocess_data import (
    DEFAULT_ARCHIVE_PATH,
    DEFAULT_OUTPUT_PATH,
    DEFAULT_SPLIT_DIR,
    load_creditcard_archive,
    preprocess,
    scale_train_validation_test_splits,
    stratified_train_validation_test_split,
    summarize,
    summarize_split,
)

raw_df = load_creditcard_archive(DEFAULT_ARCHIVE_PATH)
clean_df = preprocess(raw_df)
```

### Basic Checks

```python
print(summarize(raw_df, clean_df))
```

### Preprocessing

```python
train_df, validation_df, test_df = stratified_train_validation_test_split(clean_df)
train_df, validation_df, test_df = scale_train_validation_test_splits(
    train_df,
    validation_df,
    test_df,
)

print("Stratified train/validation/test split")
print(summarize_split("Train", train_df))
print(summarize_split("Validation", validation_df))
print(summarize_split("Test", test_df))
print("Scaled Time and Amount with StandardScaler fit on train only.")
```

### Shared Artifacts

```python
DEFAULT_OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
DEFAULT_SPLIT_DIR.mkdir(parents=True, exist_ok=True)

clean_df.to_csv(DEFAULT_OUTPUT_PATH, index=False)
train_df.to_csv(DEFAULT_SPLIT_DIR / "train.csv", index=False)
validation_df.to_csv(DEFAULT_SPLIT_DIR / "validation.csv", index=False)
test_df.to_csv(DEFAULT_SPLIT_DIR / "test.csv", index=False)

for path in [
    DEFAULT_OUTPUT_PATH,
    DEFAULT_SPLIT_DIR / "train.csv",
    DEFAULT_SPLIT_DIR / "validation.csv",
    DEFAULT_SPLIT_DIR / "test.csv",
]:
    print(path)
```
